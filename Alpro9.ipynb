```python
def kinetic_energy(mass, velocity):
    """Energi kinetik (J) dari massa (kg) dan kecepatan (m/s)."""
    return 0.5 * mass * velocity**2

def momentum(mass, velocity):
    return mass * velocity

print(f"E = {kinetic_energy(2.0, 3.0):.3f} J")
print(f"p = {momentum(2.0, 3.0):.3f} kg m/s")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 11 titik data yang sejajar dengan garis model (naik ke atas)
t = np.linspace(0.0, 5.0, 11)
# Menghitung x agar sejajar tepat pada garis model dengan batas maksimum posisi mendekati 14
x0, v0, a = 0.4, 1.2, 0.7
x = x0 + v0 * t + 0.5 * a * t**2

# Model kurva mulus
t_model = np.linspace(0.0, 5.0, 200)
x_model = x0 + v0 * t_model + 0.5 * a * t_model**2

fig, ax = plt.subplots(figsize=(6.0, 3.8))
ax.plot(t, x, "o", color="#5A2D6E", label="data (11 titik)")
ax.plot(t_model, x_model, "-", color="#3F7CAC",
        label=r"model $x_0 + v_0 t + \frac{1}{2} a t^2$")

ax.set_xlabel("Waktu, $t$ (s)", fontsize=12)
ax.set_ylabel("Posisi, $x$ (m)", fontsize=12)

# Mengatur sumbu y menjadi kelipatan genap sampai 14
ax.set_yticks(np.arange(0, 15, 2))
ax.set_ylim(-0.5, 14.5)

ax.tick_params(direction="in", top=True, right=True, labelsize=11)
ax.legend(fontsize=11)
fig.tight_layout()
fig.savefig("gerak-posisi.pdf")
print("saved gerak-posisi.pdf")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import os

# Membuat folder Data dan file contoh cooling.csv jika tidak ada
os.makedirs("Data", exist_ok=True)
cooling_file = "Data/cooling.csv"

# Selalu tulis ulang file untuk memperbarui titik data agar sejajar sempurna tanpa noise
with open(cooling_file, "w", encoding="utf-8") as f:
    f.write("time_min,temperature_C\n")
    # Menghasilkan 16 titik data dari 0 sampai 30 menit
    t_vals = np.linspace(0.0, 30.0, 16)
    T_room_sample = 25.0
    T_init_sample = 80.0
    k_sample = 0.08
    for t_val in t_vals:
        # Menghapus np.random.normal agar data sejajar sempurna dengan garis kurva
        T_val = T_room_sample + (T_init_sample - T_room_sample) * np.exp(-k_sample * t_val)
        f.write(f"{t_val:.1f},{T_val:.2f}\n")

data = np.genfromtxt(cooling_file, delimiter=",", names=True)
t, T = data["time_min"], data["temperature_C"]

rate = np.diff(T) / np.diff(t)          # C/min
i = np.argmin(rate)

print(f"samples          : {T.size}")
print(f"initial / final  : {T[0]:.2f} C / {T[-1]:.2f} C")
print(f"all rates < 0    : {np.all(rate < 0)}")
print(f"fastest cooling  : {rate[i]:.3f} C/min in [{t[i]:.0f}, {t[i+1]:.0f}] min")
print(f"slowest cooling  : {rate[-1]:.3f} C/min in [{t[-2]:.0f}, {t[-1]:.0f}] min")

T_room, k = 25.0, 0.08                  # parameter model
t_model = np.linspace(0, 30, 200)
T_model = T_room + (T[0] - T_room) * np.exp(-k * t_model)

fig, ax = plt.subplots(figsize=(6.0, 3.8))
ax.plot(t, T, "o", color="#5A2D6E", label="data (16 titik)")
ax.plot(t_model, T_model, "-", color="#3F7CAC", label="hukum pendinginan Newton")
ax.axhline(T_room, color="gray", linestyle="--", linewidth=1, label=r"$T_{\mathrm{ruang}}$")
ax.set_xlabel("Waktu, $t$ (menit)", fontsize=12)
ax.set_ylabel("Temperatur, $T$ (°C)", fontsize=12)
ax.tick_params(direction="in", top=True, right=True, labelsize=11)
ax.legend(fontsize=11)
fig.tight_layout()
fig.savefig("pendinginan.pdf")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import os

# Membuat folder Data dan file contoh wave.csv jika tidak ada
os.makedirs("Data", exist_ok=True)
wave_file = "Data/wave.csv"
if not os.path.exists(wave_file):
    with open(wave_file, "w", encoding="utf-8") as f:
        f.write("time_s,displacement_m\n")
        t_vals = np.linspace(0.0, 10.0, 500)
        # Membuat sinyal sinusoidal amplitudo 0.12 m dengan frekuensi 1 Hz
        y_vals = 0.12 * np.sin(2 * np.pi * 1.0 * t_vals)
        for tv, yv in zip(t_vals, y_vals):
            f.write(f"{tv:.4f},{yv:.6f}\n")

data = np.genfromtxt(wave_file, delimiter=",", names=True)
t, y = data["time_s"], data["displacement_m"]

amplitude = np.max(np.abs(y))
rms = np.sqrt(np.mean(y**2))
signs = np.sign(y)
signs = signs[signs != 0]               # abaikan sampel yang tepat nol
crossings = np.count_nonzero(signs[:-1] * signs[1:] < 0)

print(f"samples   = {y.size}, duration = {t[-1] - t[0]:.2f} s")
print(f"amplitude = {amplitude:.6f} m")
print(f"RMS       = {rms:.6f} m")
print(f"A/sqrt(2) = {amplitude / np.sqrt(2):.6f} m")
print(f"sign changes = {crossings}")

fig, ax = plt.subplots(figsize=(6.0, 3.4))
ax.plot(t, 100 * y, "-", color="#5A2D6E", linewidth=1.5)
ax.axhline(0, color="gray", linewidth=0.8)
ax.axhline(100 * rms, color="#3F7CAC", linestyle="--", label="RMS")
ax.set_xlabel("Waktu, $t$ (s)", fontsize=12)
ax.set_ylabel("Simpangan, $y$ (cm)", fontsize=12)
ax.tick_params(direction="in", top=True, right=True, labelsize=11)
ax.legend(fontsize=11, loc="upper right")
fig.tight_layout()
fig.savefig("gelombang.pdf")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import os

# Membuat file contoh velocity.csv dengan deret kecepatan 1.5 sampai 4.5 (kelipatan 0.5)
velocity_file = "velocity.csv"

# Selalu tulis ulang file agar datanya diperbarui sesuai permintaan
with open(velocity_file, "w", encoding="utf-8") as f:
    f.write("time_start_s,time_end_s,velocity_m_per_s\n")
    # Membuat deret kecepatan dari 1.5 sampai 4.5 dengan selang 0.5 (ada 7 titik data)
    v_vals = np.arange(1.5, 4.6, 0.5)
    for i, v_val in enumerate(v_vals):
        t_start = float(i)
        t_end = float(i + 1)
        # Menulis data tanpa menambahkan noise acak agar sejajar sempurna
        f.write(f"{t_start:.1f},{t_end:.1f},{v_val:.1f}\n")

data = np.genfromtxt(velocity_file, delimiter=",", names=True)
t_mid = 0.5 * (data["time_start_s"] + data["time_end_s"])
v = data["velocity_m_per_s"]

slope, intercept = np.polyfit(t_mid, v, 1)
print(f"a  = {slope:.4f} m/s^2")
print(f"v0 = {intercept:.4f} m/s")

fig, ax = plt.subplots(figsize=(6.0, 3.4))
ax.plot(t_mid, v, "s", color="#5A2D6E", label="hasil program C")
ax.plot(t_mid, slope * t_mid + intercept, "-", color="#3F7CAC",
        label=f"garis lurus, $a$ = {slope:.2f} m/s$^2$")
ax.set_xlabel("Waktu titik tengah interval (s)", fontsize=12)
ax.set_ylabel("Kecepatan, $v$ (m/s)", fontsize=12)
ax.tick_params(direction="in", top=True, right=True, labelsize=11)
ax.legend(fontsize=11)
fig.tight_layout()
fig.savefig("kecepatan.pdf")
```
